```python
import cv2
import matplotlib.pyplot as plt

# Change this path to your image path
image_path = r"C:\Users\admin\Desktop\1.png"

img = cv2.imread(image_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

```python
# Resize
resized = cv2.resize(img, (400, 300))

# Crop
cropped = img[50:250, 100:400]

# Grayscale
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)

# Blur
blur = cv2.GaussianBlur(img, (7, 7), 0)

# Edge detection
edges = cv2.Canny(gray, 100, 200)

plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(resized)
plt.title("Resized")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(cropped)
plt.title("Cropped")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(blur)
plt.title("Blurred")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(edges, cmap="gray")
plt.title("Edge Detection")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Drawing on the image
edited = img.copy()

cv2.rectangle(edited, (20, 20), (250, 150), (255, 0, 0), 3)
cv2.circle(edited, (300, 200), 50, (0, 255, 0), 3)
cv2.putText(edited, "OpenCV", (20, 280),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)

plt.imshow(edited)
plt.title("Edited Image")
plt.axis("off")
plt.show()
```
